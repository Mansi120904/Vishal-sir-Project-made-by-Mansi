```python
import tkinter as tk
from tkinter import messagebox
from PIL import Image, ImageTk
import random
import os

# =========================================================
# PYGAME MUSIC
# =========================================================

try:
    import pygame
    pygame.mixer.init()
    MUSIC_AVAILABLE = True
except Exception:
    MUSIC_AVAILABLE = False


# =========================================================
# MAIN WINDOW
# =========================================================

root = tk.Tk()
root.title("🎮 Game Collection")
root.geometry("700x700")
root.resizable(False, False)

BG_COLOR = "#17202A"
TITLE_COLOR = "#FFD700"


# =========================================================
# FILE PATHS
# =========================================================

IMAGE_FOLDER = "images"
MUSIC_FOLDER = "music"


# =========================================================
# MUSIC VARIABLES
# =========================================================

current_music = None


def play_music(filename):
    """
    Plays the selected music continuously.
    """
    global current_music

    if not MUSIC_AVAILABLE:
        return

    path = os.path.join(MUSIC_FOLDER, filename)

    if not os.path.exists(path):
        return

    try:
        pygame.mixer.music.stop()
        pygame.mixer.music.load(path)
        pygame.mixer.music.play(-1)
        current_music = filename
    except Exception:
        pass


def stop_music():
    """
    Stops currently playing music.
    """
    global current_music

    if MUSIC_AVAILABLE:
        try:
            pygame.mixer.music.stop()
        except Exception:
            pass

    current_music = None


# =========================================================
# BACKGROUND IMAGE FUNCTION
# =========================================================

def set_background(window, filename, width, height):
    """
    Adds a background image to a game window or main window.
    """
    path = os.path.join(IMAGE_FOLDER, filename)

    if os.path.exists(path):
        try:
            image = Image.open(path)
            image = image.resize((width, height), Image.Resampling.LANCZOS)
            photo = ImageTk.PhotoImage(image, master=window)

            background = tk.Label(window, image=photo)
            background.place(x=0, y=0, width=width, height=height)

            # Keep image reference in memory
            window.background_photo = photo
            return background
        except Exception:
            pass

    # Fallback if image fails or doesn't exist
    window.configure(bg="#17202A")
    return None


# =========================================================
# MUSIC BUTTON
# =========================================================

def create_music_button(game, music_file, x=10, y=10):
    music_state = [True]

    button = tk.Button(
        game,
        text="🔊 Music: ON",
        font=("Arial", 10, "bold"),
        bg="#2ECC71",
        fg="white",
        width=14,
        cursor="hand2"
    )
    button.place(x=x, y=y)

    def toggle_music():
        if music_state[0]:
            stop_music()
            music_state[0] = False
            button.config(text="🔇 Music: OFF", bg="#E74C3C")
        else:
            music_state[0] = True
            play_music(music_file)
            button.config(text="🔊 Music: ON", bg="#2ECC71")

    button.config(command=toggle_music)
    return button


# =========================================================
# GAME CLOSE FUNCTION
# =========================================================

def close_game(game):
    stop_music()
    game.destroy()
    # Resume main menu music when returning
    play_music("main_music.mp3")


# =========================================================
# TIC-TAC-TOE
# =========================================================

def tic_tac_toe():
    game = tk.Toplevel(root)
    game.title("❌⭕ Tic-Tac-Toe")
    game.geometry("500x650")
    game.resizable(False, False)

    set_background(game, "tictactoe_bg.jpg", 500, 650)
    play_music("tictactoe_music.mp3")
    create_music_button(game, "tictactoe_music.mp3")

    game.protocol("WM_DELETE_WINDOW", lambda: close_game(game))

    frame = tk.Frame(game, bg="#17202A")
    frame.place(x=70, y=80, width=360, height=500)

    tk.Label(
        frame, text="❌⭕ TIC-TAC-TOE",
        font=("Arial", 24, "bold"), fg="#F7DC6F", bg="#17202A"
    ).pack(pady=15)

    board = [""] * 9
    buttons = []
    player = ["X"]

    turn_label = tk.Label(
        frame, text="Player X's Turn",
        font=("Arial", 15, "bold"), fg="white", bg="#17202A"
    )
    turn_label.pack(pady=5)

    board_frame = tk.Frame(frame, bg="#17202A")
    board_frame.pack(pady=10)

    def reset_game():
        for i in range(9):
            board[i] = ""
            buttons[i].config(text="", bg="#2C3E50")
        player[0] = "X"
        turn_label.config(text="Player X's Turn")

    def check_winner():
        combinations = [
            (0, 1, 2), (3, 4, 5), (6, 7, 8),
            (0, 3, 6), (1, 4, 7), (2, 5, 8),
            (0, 4, 8), (2, 4, 6)
        ]

        for a, b, c in combinations:
            if board[a] != "" and board[a] == board[b] and board[b] == board[c]:
                buttons[a].config(bg="#27AE60")
                buttons[b].config(bg="#27AE60")
                buttons[c].config(bg="#27AE60")
                messagebox.showinfo("Winner", f"🎉 Player {board[a]} Wins!")
                reset_game()
                return True

        if "" not in board:
            messagebox.showinfo("Game Over", "🤝 It's a Draw!")
            reset_game()
            return True

        return False

    def play(index):
        if board[index] != "":
            return

        board[index] = player[0]
        buttons[index].config(
            text=player[0],
            fg="#E74C3C" if player[0] == "X" else "#3498DB"
        )

        if check_winner():
            return

        player[0] = "O" if player[0] == "X" else "X"
        turn_label.config(text=f"Player {player[0]}'s Turn")

    for i in range(9):
        button = tk.Button(
            board_frame, text="", font=("Arial", 22, "bold"),
            width=4, height=2, bg="#2C3E50", fg="white",
            command=lambda i=i: play(i)
        )
        button.grid(row=i // 3, column=i % 3, padx=3, pady=3)
        buttons.append(button)

    tk.Button(
        frame, text="🔄 NEW GAME", font=("Arial", 13, "bold"),
        bg="#E67E22", fg="white", width=15, command=reset_game
    ).pack(pady=10)


# =========================================================
# SNAKE AND LADDER
# =========================================================

def snake_and_ladder():
    game = tk.Toplevel(root)
    game.title("🎲 Snake and Ladder")
    game.geometry("650x750")
    game.resizable(False, False)

    set_background(game, "snake_ladder_bg.jpg", 650, 750)
    play_music("snake_ladder_music.mp3")
    create_music_button(game, "snake_ladder_music.mp3")

    game.protocol("WM_DELETE_WINDOW", lambda: close_game(game))

    frame = tk.Frame(game, bg="#17202A")
    frame.place(x=25, y=65, width=600, height=660)

    tk.Label(
        frame, text="🎲 SNAKE & LADDER",
        font=("Arial", 25, "bold"), fg="#F7DC6F", bg="#17202A"
    ).pack(pady=8)

    position = [1]
    snakes = {99: 54, 95: 75, 92: 73, 64: 60, 62: 19, 56: 53, 49: 11, 47: 26}
    ladders = {4: 25, 9: 31, 21: 42, 28: 55, 36: 44, 51: 67, 71: 91, 80: 100}

    board_frame = tk.Frame(frame, bg="#17202A")
    board_frame.pack()

    cells = {}
    for row in range(10):
        for col in range(10):
            number = 100 - row * 10 - (col if row % 2 == 0 else 9 - col)
            cell = tk.Label(
                board_frame, text=str(number), font=("Arial", 8, "bold"),
                width=5, height=2, relief="solid", bd=1, bg="#F8F9F9"
            )
            cell.grid(row=row, column=col)
            cells[number] = cell

    position_label = tk.Label(
        frame, text="Position: 1", font=("Arial", 14, "bold"),
        fg="white", bg="#17202A"
    )
    position_label.pack(pady=5)

    dice_label = tk.Label(
        frame, text="🎲", font=("Arial", 30), fg="white", bg="#17202A"
    )
    dice_label.pack()

    def update_board():
        for number in cells:
            cells[number].config(text=str(number), bg="#F8F9F9")
        for start in ladders:
            cells[start].config(text=f"🪜 {start}", bg="#ABEBC6")
        for start in snakes:
            cells[start].config(text=f"🐍 {start}", bg="#F5B7B1")
        cells[position[0]].config(text=f"🔴 {position[0]}", bg="#F9E79F")

    def roll_dice():
        dice = random.randint(1, 6)
        dice_label.config(text=f"🎲 {dice}")

        if position[0] + dice <= 100:
            position[0] += dice

        if position[0] in ladders:
            position[0] = ladders[position[0]]
        elif position[0] in snakes:
            position[0] = snakes[position[0]]

        position_label.config(text=f"Position: {position[0]}")
        update_board()

        if position[0] == 100:
            messagebox.showinfo("Winner", "🎉 Congratulations!\nYou reached 100!")

    def reset():
        position[0] = 1
        position_label.config(text="Position: 1")
        dice_label.config(text="🎲")
        update_board()

    tk.Button(
        frame, text="🎲 ROLL DICE", font=("Arial", 14, "bold"),
        bg="#3498DB", fg="white", width=18, command=roll_dice
    ).pack(pady=5)

    tk.Button(
        frame, text="🔄 NEW GAME", font=("Arial", 13),
        bg="#E67E22", fg="white", width=18, command=reset
    ).pack()

    update_board()


# =========================================================
# ROCK PAPER SCISSORS
# =========================================================

def rock_paper_scissors():
    game = tk.Toplevel(root)
    game.title("✊ Rock Paper Scissors")
    game.geometry("500x650")
    game.resizable(False, False)

    set_background(game, "rps_bg.jpg", 500, 650)
    play_music("rps_music.mp3")
    create_music_button(game, "rps_music.mp3")

    game.protocol("WM_DELETE_WINDOW", lambda: close_game(game))

    frame = tk.Frame(game, bg="#1B2631")
    frame.place(x=50, y=80, width=400, height=540)

    tk.Label(
        frame, text="✊✋✌️ ROCK PAPER SCISSORS",
        font=("Arial", 19, "bold"), fg="#F7DC6F", bg="#1B2631"
    ).pack(pady=20)

    result = tk.Label(
        frame, text="Choose your move!",
        font=("Arial", 18, "bold"), fg="white", bg="#1B2631"
    )
    result.pack(pady=15)

    computer_label = tk.Label(
        frame, text="Computer: ❓",
        font=("Arial", 15), fg="#AED6F1", bg="#1B2631"
    )
    computer_label.pack(pady=8)

    score = {"player": 0, "computer": 0, "draw": 0}

    score_label = tk.Label(
        frame, text="You: 0     Computer: 0     Draw: 0",
        font=("Arial", 13, "bold"), fg="white", bg="#1B2631"
    )
    score_label.pack(pady=15)

    choices = ["Rock", "Paper", "Scissors"]

    def play(player_choice):
        computer_choice = random.choice(choices)
        computer_label.config(text=f"Computer: {computer_choice}")

        if player_choice == computer_choice:
            score["draw"] += 1
            result.config(text="🤝 DRAW!", fg="#F1C40F")
        elif (
            (player_choice == "Rock" and computer_choice == "Scissors") or
            (player_choice == "Paper" and computer_choice == "Rock") or
            (player_choice == "Scissors" and computer_choice == "Paper")
        ):
            score["player"] += 1
            result.config(text="🎉 YOU WIN!", fg="#2ECC71")
        else:
            score["computer"] += 1
            result.config(text="😔 COMPUTER WINS!", fg="#E74C3C")

        score_label.config(
            text=f"You: {score['player']}     Computer: {score['computer']}     Draw: {score['draw']}"
        )

    tk.Button(
        frame, text="✊ ROCK", font=("Arial", 14, "bold"),
        bg="#E74C3C", fg="white", width=18, command=lambda: play("Rock")
    ).pack(pady=6)

    tk.Button(
        frame, text="✋ PAPER", font=("Arial", 14, "bold"),
        bg="#3498DB", fg="white", width=18, command=lambda: play("Paper")
    ).pack(pady=6)

    tk.Button(
        frame, text="✌️ SCISSORS", font=("Arial", 14, "bold"),
        bg="#9B59B6", fg="white", width=18, command=lambda: play("Scissors")
    ).pack(pady=6)


# =========================================================
# NUMBER GUESSING
# =========================================================

def number_guessing():
    game = tk.Toplevel(root)
    game.title("🔢 Number Guessing")
    game.geometry("500x550")
    game.resizable(False, False)

    set_background(game, "number_guessing_bg.jpg", 500, 550)
    play_music("number_guessing_music.mp3")
    create_music_button(game, "number_guessing_music.mp3")

    game.protocol("WM_DELETE_WINDOW", lambda: close_game(game))

    frame = tk.Frame(game, bg="#1B2631")
    frame.place(x=50, y=75, width=400, height=430)

    tk.Label(
        frame, text="🔢 NUMBER GUESSING",
        font=("Arial", 24, "bold"), fg="#F7DC6F", bg="#1B2631"
    ).pack(pady=25)

    tk.Label(
        frame, text="Guess a number between 1 and 100",
        font=("Arial", 14), fg="white", bg="#1B2631"
    ).pack(pady=10)

    number = [random.randint(1, 100)]
    attempts = [0]

    entry = tk.Entry(frame, font=("Arial", 20), justify="center", width=10)
    entry.pack(pady=15)

    result = tk.Label(
        frame, text="Good luck! 🎯",
        font=("Arial", 15, "bold"), fg="#AED6F1", bg="#1B2631"
    )
    result.pack(pady=10)

    attempts_label = tk.Label(
        frame, text="Attempts: 0", font=("Arial", 13), fg="white", bg="#1B2631"
    )
    attempts_label.pack(pady=5)

    def reset():
        number[0] = random.randint(1, 100)
        attempts[0] = 0
        entry.delete(0, tk.END)
        result.config(text="New number generated! 🎯")
        attempts_label.config(text="Attempts: 0")

    def guess_number():
        try:
            guess = int(entry.get())
        except ValueError:
            result.config(text="Please enter a number!")
            return

        attempts[0] += 1
        attempts_label.config(text=f"Attempts: {attempts[0]}")

        if guess < number[0]:
            result.config(text="⬆️ Too Low!")
        elif guess > number[0]:
            result.config(text="⬇️ Too High!")
        else:
            messagebox.showinfo("Congratulations!", f"🎉 Correct!\nAttempts: {attempts[0]}")
            reset()

    tk.Button(
        frame, text="🎯 GUESS", font=("Arial", 14, "bold"),
        bg="#27AE60", fg="white", width=15, command=guess_number
    ).pack(pady=8)

    tk.Button(
        frame, text="🔄 NEW GAME", font=("Arial", 13),
        bg="#E67E22", fg="white", width=15, command=reset
    ).pack(pady=5)


# =========================================================
# MEMORY GAME
# =========================================================

def memory_game():
    game = tk.Toplevel(root)
    game.title("🧠 Memory Game")
    game.geometry("600x700")
    game.resizable(False, False)

    set_background(game, "memory_bg.jpg", 600, 700)
    play_music("memory_music.mp3")
    create_music_button(game, "memory_music.mp3")

    game.protocol("WM_DELETE_WINDOW", lambda: close_game(game))

    frame = tk.Frame(game, bg="#17202A")
    frame.place(x=50, y=70, width=500, height=600)

    tk.Label(
        frame, text="🧠 MEMORY GAME",
        font=("Arial", 26, "bold"), fg="#F7DC6F", bg="#17202A"
    ).pack(pady=10)

    moves_label = tk.Label(
        frame, text="Moves: 0", font=("Arial", 15, "bold"),
        fg="white", bg="#17202A"
    )
    moves_label.pack(pady=5)

    cards_frame = tk.Frame(frame, bg="#17202A")
    cards_frame.pack(pady=10)

    emojis = ["🍎", "🍌", "🍇", "🍉", "🍓", "🥝", "🍒", "🥭"]
    cards = emojis * 2
    random.shuffle(cards)

    buttons = []
    revealed = []
    matched = []
    moves = [0]

    def reset():
        random.shuffle(cards)
        moves[0] = 0
        revealed.clear()
        matched.clear()
        moves_label.config(text="Moves: 0")
        for button in buttons:
            button.config(text="?", state="normal", bg="#3498DB")

    def select_card(index):
        if index in matched or index in revealed or len(revealed) == 2:
            return

        buttons[index].config(text=cards[index], bg="#F7DC6F")
        revealed.append(index)

        if len(revealed) == 2:
            moves[0] += 1
            moves_label.config(text=f"Moves: {moves[0]}")
            game.after(700, check_match)

    def check_match():
        if len(revealed) != 2:
            return

        first, second = revealed[0], revealed[1]

        if cards[first] == cards[second]:
            matched.extend([first, second])
            buttons[first].config(bg="#2ECC71")
            buttons[second].config(bg="#2ECC71")
        else:
            buttons[first].config(text="?", bg="#3498DB")
            buttons[second].config(text="?", bg="#3498DB")

        revealed.clear()

        if len(matched) == 16:
            messagebox.showinfo("Congratulations!", f"🎉 You matched all cards!\nMoves: {moves[0]}")

    for i in range(16):
        button = tk.Button(
            cards_frame, text="?", font=("Arial", 22, "bold"),
            width=5, height=2, bg="#3498DB", fg="white",
            command=lambda i=i: select_card(i)
        )
        button.grid(row=i // 4, column=i % 4, padx=5, pady=5)
        buttons.append(button)

    tk.Button(
        frame, text="🔄 NEW GAME", font=("Arial", 13, "bold"),
        bg="#E67E22", fg="white", width=15, command=reset
    ).pack(pady=10)


# =========================================================
# SUDOKU
# =========================================================

def sudoku():
    game = tk.Toplevel(root)
    game.title("🔢 Sudoku")
    game.geometry("650x750")
    game.resizable(False, False)

    set_background(game, "sudoku_bg.jpg", 650, 750)
    play_music("sudoku_music.mp3")
    create_music_button(game, "sudoku_music.mp3")

    game.protocol("WM_DELETE_WINDOW", lambda: close_game(game))

    frame = tk.Frame(game, bg="#17202A")
    frame.place(x=25, y=65, width=600, height=665)

    tk.Label(
        frame, text="🔢 SUDOKU", font=("Arial", 28, "bold"),
        fg="#F7DC6F", bg="#17202A"
    ).pack(pady=8)

    difficulty = tk.StringVar(value="Easy")

    difficulty_frame = tk.Frame(frame, bg="#17202A")
    difficulty_frame.pack(pady=5)

    tk.Label(
        difficulty_frame, text="Difficulty:", font=("Arial", 12, "bold"),
        fg="white", bg="#17202A"
    ).pack(side="left", padx=5)

    difficulty_menu = tk.OptionMenu(
        difficulty_frame, difficulty, "Easy", "Medium", "Hard"
    )
    difficulty_menu.config(font=("Arial", 11, "bold"), width=10)
    difficulty_menu.pack(side="left")

    board_frame = tk.Frame(frame, bg="#17202A")
    board_frame.pack(pady=5)

    entries = []
    puzzle = []
    solution = []

    def is_valid(board, row, col, num):
        for i in range(9):
            if board[row][i] == num or board[i][col] == num:
                return False

        start_row, start_col = (row // 3) * 3, (col // 3) * 3
        for i in range(start_row, start_row + 3):
            for j in range(start_col, start_col + 3):
                if board[i][j] == num:
                    return False
        return True

    def solve_board(board):
        for row in range(9):
            for col in range(9):
                if board[row][col] == 0:
                    numbers = list(range(1, 10))
                    random.shuffle(numbers)
                    for num in numbers:
                        if is_valid(board, row, col, num):
                            board[row][col] = num
                            if solve_board(board):
                                return True
                            board[row][col] = 0
                    return False
        return True

    def generate_sudoku():
        nonlocal puzzle, solution
        solution = [[0 for _ in range(9)] for _ in range(9)]
        solve_board(solution)

        puzzle = [row[:] for row in solution]
        remove_count = 35 if difficulty.get() == "Easy" else (45 if difficulty.get() == "Medium" else 55)

        positions = [(r, c) for r in range(9) for c in range(9)]
        random.shuffle(positions)

        for i in range(remove_count):
            r, c = positions[i]
            puzzle[r][c] = 0

    def display_puzzle():
        for row in range(9):
            for col in range(9):
                value = puzzle[row][col]
                entry = entries[row][col]
                entry.config(state="normal")
                entry.delete(0, tk.END)

                if value != 0:
                    entry.insert(0, str(value))
                    entry.config(state="disabled", disabledforeground="#D35400")
                else:
                    entry.config(bg="white", fg="black")

    def new_game():
        generate_sudoku()
        display_puzzle()

    def check_answer():
        correct = True
        for row in range(9):
            for col in range(9):
                value = entries[row][col].get()
                if value == "":
                    correct = False
                    entries[row][col].config(bg="#F5B7B1")
                else:
                    try:
                        if int(value) != solution[row][col]:
                            correct = False
                            entries[row][col].config(bg="#F5B7B1")
                        else:
                            entries[row][col].config(bg="#ABEBC6")
                    except ValueError:
                        correct = False
                        entries[row][col].config(bg="#F5B7B1")

        if correct:
            messagebox.showinfo("Sudoku", "🎉 Congratulations!\nYou solved the Sudoku!")

    def clear_answer():
        for row in range(9):
            for col in range(9):
                if puzzle[row][col] == 0:
                    entries[row][col].delete(0, tk.END)
                    entries[row][col].config(bg="white")

    def solve_current():
        for row in range(9):
            for col in range(9):
                entries[row][col].config(state="normal")
                entries[row][col].delete(0, tk.END)
                entries[row][col].insert(0, str(solution[row][col]))
                entries[row][col].config(bg="#ABEBC6")

    # Construct Grid
    for row in range(9):
        row_entries = []
        for col in range(9):
            pad_x = (2, 5) if (col + 1) % 3 == 0 and col < 8 else (2, 2)
            pad_y = (2, 5) if (row + 1) % 3 == 0 and row < 8 else (2, 2)

            entry = tk.Entry(
                board_frame, font=("Arial", 16, "bold"), width=2,
                justify="center", relief="solid", bd=1
            )
            entry.grid(row=row, column=col, padx=pad_x, pady=pad_y)
            row_entries.append(entry)
        entries.append(row_entries)

    btn_frame = tk.Frame(frame, bg="#17202A")
    btn_frame.pack(pady=10)

    tk.Button(
        btn_frame, text="🔄 New", font=("Arial", 11, "bold"),
        bg="#E67E22", fg="white", width=8, command=new_game
    ).grid(row=0, column=0, padx=5)

    tk.Button(
        btn_frame, text="✅ Check", font=("Arial", 11, "bold"),
        bg="#2ECC71", fg="white", width=8, command=check_answer
    ).grid(row=0, column=1, padx=5)

    tk.Button(
        btn_frame, text="🧹 Clear", font=("Arial", 11, "bold"),
        bg="#3498DB", fg="white", width=8, command=clear_answer
    ).grid(row=0, column=2, padx=5)

    tk.Button(
        btn_frame, text="💡 Solve", font=("Arial", 11, "bold"),
        bg="#9B59B6", fg="white", width=8, command=solve_current
    ).grid(row=0, column=3, padx=5)

    new_game()


# =========================================================
# MAIN MENU UI (MATCHES IMAGE 2)
# =========================================================

# Set Main Window Background
set_background(root, "main_bg.jpg", 700, 700)

# Title Banner
title_label = tk.Label(
    root,
    text="🎮  GAME COLLECTION  🎮",
    font=("Arial", 22, "bold"),
    fg=TITLE_COLOR,
    bg="#0B131E",
    padx=20,
    pady=10,
    relief="solid",
    bd=2
)
title_label.pack(pady=(25, 10))

# Subtitle Banner
subtitle_label = tk.Label(
    root,
    text="Choose a game and have fun!",
    font=("Arial", 13, "bold"),
    fg="white",
    bg="#0B131E",
    padx=15,
    pady=5
)
subtitle_label.pack(pady=(0, 20))

# Menu Buttons Setup
button_style = {
    "font": ("Arial", 13, "bold"),
    "fg": "white",
    "width": 26,
    "height": 1,
    "bd": 2,
    "relief": "raised",
    "cursor": "hand2"
}

tk.Button(
    root, text="❌ ⭕   TIC-TAC-TOE", bg="#E74C3C",
    command=tic_tac_toe, **button_style
).pack(pady=6)

tk.Button(
    root, text="🎲   SNAKE & LADDER", bg="#2980B9",
    command=snake_and_ladder, **button_style
).pack(pady=6)

tk.Button(
    root, text="✊   ROCK PAPER SCISSORS", bg="#8E44AD",
    command=rock_paper_scissors, **button_style
).pack(pady=6)

tk.Button(
    root, text="🔢   NUMBER GUESSING", bg="#27AE60",
    command=number_guessing, **button_style
).pack(pady=6)

tk.Button(
    root, text="🧠   MEMORY GAME", bg="#9B59B6",
    command=memory_game, **button_style
).pack(pady=6)

tk.Button(
    root, text="🔢   SUDOKU", bg="#16A085",
    command=sudoku, **button_style
).pack(pady=6)

tk.Button(
    root, text="🚪   EXIT", bg="#5D6D7E",
    command=root.quit, **button_style
).pack(pady=(6, 25))

# Footer Tag
footer_label = tk.Label(
    root,
    text="Made with Python 🐍",
    font=("Arial", 10, "bold"),
    fg="white",
    bg="#0B131E",
    padx=10,
    pady=3
)
footer_label.pack(side="bottom", pady=15)

# Start background music for main window
play_music("main_music.mp3")

root.mainloop()
```

```python
import tkinter as tk
from tkinter import messagebox
from PIL import Image, ImageTk
import random
import os

# =========================================================
# PYGAME MUSIC
# =========================================================

try:
    import pygame
    pygame.mixer.init()
    MUSIC_AVAILABLE = True
except Exception:
    MUSIC_AVAILABLE = False


# =========================================================
# MAIN WINDOW
# =========================================================

root = tk.Tk()
root.title("🎮 Game Collection")
root.geometry("700x700")
root.resizable(False, False)

BG_COLOR = "#17202A"
TITLE_COLOR = "#FFD700"


# =========================================================
# FILE PATHS (Set to current folder where files exist)
# =========================================================

IMAGE_FOLDER = "."
MUSIC_FOLDER = "."


# =========================================================
# MUSIC VARIABLES
# =========================================================

current_music = None


def play_music(filename):
    """
    Plays the selected music continuously.
    """
    global current_music

    if not MUSIC_AVAILABLE:
        return

    path = os.path.join(MUSIC_FOLDER, filename)

    if not os.path.exists(path):
        return

    try:
        pygame.mixer.music.stop()
        pygame.mixer.music.load(path)
        pygame.mixer.music.play(-1)
        current_music = filename
    except Exception:
        pass


def stop_music():
    """
    Stops currently playing music.
    """
    global current_music

    if MUSIC_AVAILABLE:
        try:
            pygame.mixer.music.stop()
        except Exception:
            pass

    current_music = None


# =========================================================
# BACKGROUND IMAGE FUNCTION
# =========================================================

def set_background(window, filename, width, height):
    """
    Adds a background image to a game window or main window.
    """
    path = os.path.join(IMAGE_FOLDER, filename)

    if os.path.exists(path):
        try:
            image = Image.open(path)
            image = image.resize((width, height), Image.Resampling.LANCZOS)
            photo = ImageTk.PhotoImage(image, master=window)

            background = tk.Label(window, image=photo)
            background.place(x=0, y=0, width=width, height=height)

            # Keep image reference in memory
            window.background_photo = photo
            return background
        except Exception:
            pass

    # Fallback if image fails or doesn't exist
    window.configure(bg="#17202A")
    return None


# =========================================================
# MUSIC BUTTON
# =========================================================

def create_music_button(game, music_file, x=10, y=10):
    music_state = [True]

    button = tk.Button(
        game,
        text="🔊 Music: ON",
        font=("Arial", 10, "bold"),
        bg="#2ECC71",
        fg="white",
        width=14,
        cursor="hand2"
    )
    button.place(x=x, y=y)

    def toggle_music():
        if music_state[0]:
            stop_music()
            music_state[0] = False
            button.config(text="🔇 Music: OFF", bg="#E74C3C")
        else:
            music_state[0] = True
            play_music(music_file)
            button.config(text="🔊 Music: ON", bg="#2ECC71")

    button.config(command=toggle_music)
    return button


# =========================================================
# GAME CLOSE FUNCTION
# =========================================================

def close_game(game):
    stop_music()
    game.destroy()
    # Resume main menu music when returning
    play_music("main_music.mp3")


# =========================================================
# TIC-TAC-TOE
# =========================================================

def tic_tac_toe():
    game = tk.Toplevel(root)
    game.title("❌⭕ Tic-Tac-Toe")
    game.geometry("500x650")
    game.resizable(False, False)

    set_background(game, "tictactoe_bg.jpg", 500, 650)
    play_music("main_music.mp3")
    create_music_button(game, "main_music.mp3")

    game.protocol("WM_DELETE_WINDOW", lambda: close_game(game))

    frame = tk.Frame(game, bg="#17202A")
    frame.place(x=70, y=80, width=360, height=500)

    tk.Label(
        frame, text="❌⭕ TIC-TAC-TOE",
        font=("Arial", 24, "bold"), fg="#F7DC6F", bg="#17202A"
    ).pack(pady=15)

    board = [""] * 9
    buttons = []
    player = ["X"]

    turn_label = tk.Label(
        frame, text="Player X's Turn",
        font=("Arial", 15, "bold"), fg="white", bg="#17202A"
    )
    turn_label.pack(pady=5)

    board_frame = tk.Frame(frame, bg="#17202A")
    board_frame.pack(pady=10)

    def reset_game():
        for i in range(9):
            board[i] = ""
            buttons[i].config(text="", bg="#2C3E50")
        player[0] = "X"
        turn_label.config(text="Player X's Turn")

    def check_winner():
        combinations = [
            (0, 1, 2), (3, 4, 5), (6, 7, 8),
            (0, 3, 6), (1, 4, 7), (2, 5, 8),
            (0, 4, 8), (2, 4, 6)
        ]

        for a, b, c in combinations:
            if board[a] != "" and board[a] == board[b] and board[b] == board[c]:
                buttons[a].config(bg="#27AE60")
                buttons[b].config(bg="#27AE60")
                buttons[c].config(bg="#27AE60")
                messagebox.showinfo("Winner", f"🎉 Player {board[a]} Wins!")
                reset_game()
                return True

        if "" not in board:
            messagebox.showinfo("Game Over", "🤝 It's a Draw!")
            reset_game()
            return True

        return False

    def play(index):
        if board[index] != "":
            return

        board[index] = player[0]
        buttons[index].config(
            text=player[0],
            fg="#E74C3C" if player[0] == "X" else "#3498DB"
        )

        if check_winner():
            return

        player[0] = "O" if player[0] == "X" else "X"
        turn_label.config(text=f"Player {player[0]}'s Turn")

    for i in range(9):
        button = tk.Button(
            board_frame, text="", font=("Arial", 22, "bold"),
            width=4, height=2, bg="#2C3E50", fg="white",
            command=lambda i=i: play(i)
        )
        button.grid(row=i // 3, column=i % 3, padx=3, pady=3)
        buttons.append(button)

    tk.Button(
        frame, text="🔄 NEW GAME", font=("Arial", 13, "bold"),
        bg="#E67E22", fg="white", width=15, command=reset_game
    ).pack(pady=10)


# =========================================================
# SNAKE AND LADDER
# =========================================================

def snake_and_ladder():
    game = tk.Toplevel(root)
    game.title("🎲 Snake and Ladder")
    game.geometry("650x750")
    game.resizable(False, False)

    set_background(game, "snake_ladder_bg.jpg", 650, 750)
    play_music("snake_ladder_music.mp3")
    create_music_button(game, "snake_ladder_music.mp3")

    game.protocol("WM_DELETE_WINDOW", lambda: close_game(game))

    frame = tk.Frame(game, bg="#17202A")
    frame.place(x=25, y=65, width=600, height=660)

    tk.Label(
        frame, text="🎲 SNAKE & LADDER",
        font=("Arial", 25, "bold"), fg="#F7DC6F", bg="#17202A"
    ).pack(pady=8)

    position = [1]
    snakes = {99: 54, 95: 75, 92: 73, 64: 60, 62: 19, 56: 53, 49: 11, 47: 26}
    ladders = {4: 25, 9: 31, 21: 42, 28: 55, 36: 44, 51: 67, 71: 91, 80: 100}

    board_frame = tk.Frame(frame, bg="#17202A")
    board_frame.pack()

    cells = {}
    for row in range(10):
        for col in range(10):
            number = 100 - row * 10 - (col if row % 2 == 0 else 9 - col)
            cell = tk.Label(
                board_frame, text=str(number), font=("Arial", 8, "bold"),
                width=5, height=2, relief="solid", bd=1, bg="#F8F9F9"
            )
            cell.grid(row=row, column=col)
            cells[number] = cell

    position_label = tk.Label(
        frame, text="Position: 1", font=("Arial", 14, "bold"),
        fg="white", bg="#17202A"
    )
    position_label.pack(pady=5)

    dice_label = tk.Label(
        frame, text="🎲", font=("Arial", 30), fg="white", bg="#17202A"
    )
    dice_label.pack()

    def update_board():
        for number in cells:
            cells[number].config(text=str(number), bg="#F8F9F9")
        for start in ladders:
            cells[start].config(text=f"🪜 {start}", bg="#ABEBC6")
        for start in snakes:
            cells[start].config(text=f"🐍 {start}", bg="#F5B7B1")
        cells[position[0]].config(text=f"🔴 {position[0]}", bg="#F9E79F")

    def roll_dice():
        dice = random.randint(1, 6)
        dice_label.config(text=f"🎲 {dice}")

        if position[0] + dice <= 100:
            position[0] += dice

        if position[0] in ladders:
            position[0] = ladders[position[0]]
        elif position[0] in snakes:
            position[0] = snakes[position[0]]

        position_label.config(text=f"Position: {position[0]}")
        update_board()

        if position[0] == 100:
            messagebox.showinfo("Winner", "🎉 Congratulations!\nYou reached 100!")

    def reset():
        position[0] = 1
        position_label.config(text="Position: 1")
        dice_label.config(text="🎲")
        update_board()

    tk.Button(
        frame, text="🎲 ROLL DICE", font=("Arial", 14, "bold"),
        bg="#3498DB", fg="white", width=18, command=roll_dice
    ).pack(pady=5)

    tk.Button(
        frame, text="🔄 NEW GAME", font=("Arial", 13),
        bg="#E67E22", fg="white", width=18, command=reset
    ).pack()

    update_board()


# =========================================================
# ROCK PAPER SCISSORS
# =========================================================

def rock_paper_scissors():
    game = tk.Toplevel(root)
    game.title("✊ Rock Paper Scissors")
    game.geometry("500x650")
    game.resizable(False, False)

    set_background(game, "rps_bg.jpg", 500, 650)
    play_music("main_music.mp3")
    create_music_button(game, "main_music.mp3")

    game.protocol("WM_DELETE_WINDOW", lambda: close_game(game))

    frame = tk.Frame(game, bg="#1B2631")
    frame.place(x=50, y=80, width=400, height=540)

    tk.Label(
        frame, text="✊✋✌️ ROCK PAPER SCISSORS",
        font=("Arial", 19, "bold"), fg="#F7DC6F", bg="#1B2631"
    ).pack(pady=20)

    result = tk.Label(
        frame, text="Choose your move!",
        font=("Arial", 18, "bold"), fg="white", bg="#1B2631"
    )
    result.pack(pady=15)

    computer_label = tk.Label(
        frame, text="Computer: ❓",
        font=("Arial", 15), fg="#AED6F1", bg="#1B2631"
    )
    computer_label.pack(pady=8)

    score = {"player": 0, "computer": 0, "draw": 0}

    score_label = tk.Label(
        frame, text="You: 0     Computer: 0     Draw: 0",
        font=("Arial", 13, "bold"), fg="white", bg="#1B2631"
    )
    score_label.pack(pady=15)

    choices = ["Rock", "Paper", "Scissors"]

    def play(player_choice):
        computer_choice = random.choice(choices)
        computer_label.config(text=f"Computer: {computer_choice}")

        if player_choice == computer_choice:
            score["draw"] += 1
            result.config(text="🤝 DRAW!", fg="#F1C40F")
        elif (
            (player_choice == "Rock" and computer_choice == "Scissors") or
            (player_choice == "Paper" and computer_choice == "Rock") or
            (player_choice == "Scissors" and computer_choice == "Paper")
        ):
            score["player"] += 1
            result.config(text="🎉 YOU WIN!", fg="#2ECC71")
        else:
            score["computer"] += 1
            result.config(text="😔 COMPUTER WINS!", fg="#E74C3C")

        score_label.config(
            text=f"You: {score['player']}     Computer: {score['computer']}     Draw: {score['draw']}"
        )

    tk.Button(
        frame, text="✊ ROCK", font=("Arial", 14, "bold"),
        bg="#E74C3C", fg="white", width=18, command=lambda: play("Rock")
    ).pack(pady=6)

    tk.Button(
        frame, text="✋ PAPER", font=("Arial", 14, "bold"),
        bg="#3498DB", fg="white", width=18, command=lambda: play("Paper")
    ).pack(pady=6)

    tk.Button(
        frame, text="✌️ SCISSORS", font=("Arial", 14, "bold"),
        bg="#9B59B6", fg="white", width=18, command=lambda: play("Scissors")
    ).pack(pady=6)


# =========================================================
# NUMBER GUESSING
# =========================================================

def number_guessing():
    game = tk.Toplevel(root)
    game.title("🔢 Number Guessing")
    game.geometry("500x550")
    game.resizable(False, False)

    set_background(game, "number_guessing_bg.jpg", 500, 550)
    play_music("main_music.mp3")
    create_music_button(game, "main_music.mp3")

    game.protocol("WM_DELETE_WINDOW", lambda: close_game(game))

    frame = tk.Frame(game, bg="#1B2631")
    frame.place(x=50, y=75, width=400, height=430)

    tk.Label(
        frame, text="🔢 NUMBER GUESSING",
        font=("Arial", 24, "bold"), fg="#F7DC6F", bg="#1B2631"
    ).pack(pady=25)

    tk.Label(
        frame, text="Guess a number between 1 and 100",
        font=("Arial", 14), fg="white", bg="#1B2631"
    ).pack(pady=10)

    number = [random.randint(1, 100)]
    attempts = [0]

    entry = tk.Entry(frame, font=("Arial", 20), justify="center", width=10)
    entry.pack(pady=15)

    result = tk.Label(
        frame, text="Good luck! 🎯",
        font=("Arial", 15, "bold"), fg="#AED6F1", bg="#1B2631"
    )
    result.pack(pady=10)

    attempts_label = tk.Label(
        frame, text="Attempts: 0", font=("Arial", 13), fg="white", bg="#1B2631"
    )
    attempts_label.pack(pady=5)

    def reset():
        number[0] = random.randint(1, 100)
        attempts[0] = 0
        entry.delete(0, tk.END)
        result.config(text="New number generated! 🎯")
        attempts_label.config(text="Attempts: 0")

    def guess_number():
        try:
            guess = int(entry.get())
        except ValueError:
            result.config(text="Please enter a number!")
            return

        attempts[0] += 1
        attempts_label.config(text=f"Attempts: {attempts[0]}")

        if guess < number[0]:
            result.config(text="⬆️ Too Low!")
        elif guess > number[0]:
            result.config(text="⬇️ Too High!")
        else:
            messagebox.showinfo("Congratulations!", f"🎉 Correct!\nAttempts: {attempts[0]}")
            reset()

    tk.Button(
        frame, text="🎯 GUESS", font=("Arial", 14, "bold"),
        bg="#27AE60", fg="white", width=15, command=guess_number
    ).pack(pady=8)

    tk.Button(
        frame, text="🔄 NEW GAME", font=("Arial", 13),
        bg="#E67E22", fg="white", width=15, command=reset
    ).pack(pady=5)


# =========================================================
# MEMORY GAME
# =========================================================

def memory_game():
    game = tk.Toplevel(root)
    game.title("🧠 Memory Game")
    game.geometry("600x700")
    game.resizable(False, False)

    set_background(game, "memory_bg.png", 600, 700) # Updated filename extension to .png
    play_music("main_music.mp3")
    create_music_button(game, "main_music.mp3")

    game.protocol("WM_DELETE_WINDOW", lambda: close_game(game))

    frame = tk.Frame(game, bg="#17202A")
    frame.place(x=50, y=70, width=500, height=600)

    tk.Label(
        frame, text="🧠 MEMORY GAME",
        font=("Arial", 26, "bold"), fg="#F7DC6F", bg="#17202A"
    ).pack(pady=10)

    moves_label = tk.Label(
        frame, text="Moves: 0", font=("Arial", 15, "bold"),
        fg="white", bg="#17202A"
    )
    moves_label.pack(pady=5)

    cards_frame = tk.Frame(frame, bg="#17202A")
    cards_frame.pack(pady=10)

    emojis = ["🍎", "🍌", "🍇", "🍉", "🍓", "🥝", "🍒", "🥭"]
    cards = emojis * 2
    random.shuffle(cards)

    buttons = []
    revealed = []
    matched = []
    moves = [0]

    def reset():
        random.shuffle(cards)
        moves[0] = 0
        revealed.clear()
        matched.clear()
        moves_label.config(text="Moves: 0")
        for button in buttons:
            button.config(text="?", state="normal", bg="#3498DB")

    def select_card(index):
        if index in matched or index in revealed or len(revealed) == 2:
            return

        buttons[index].config(text=cards[index], bg="#F7DC6F")
        revealed.append(index)

        if len(revealed) == 2:
            moves[0] += 1
            moves_label.config(text=f"Moves: {moves[0]}")
            game.after(700, check_match)

    def check_match():
        if len(revealed) != 2:
            return

        first, second = revealed[0], revealed[1]

        if cards[first] == cards[second]:
            matched.extend([first, second])
            buttons[first].config(bg="#2ECC71")
            buttons[second].config(bg="#2ECC71")
        else:
            buttons[first].config(text="?", bg="#3498DB")
            buttons[second].config(text="?", bg="#3498DB")

        revealed.clear()

        if len(matched) == 16:
            messagebox.showinfo("Congratulations!", f"🎉 You matched all cards!\nMoves: {moves[0]}")

    for i in range(16):
        button = tk.Button(
            cards_frame, text="?", font=("Arial", 22, "bold"),
            width=5, height=2, bg="#3498DB", fg="white",
            command=lambda i=i: select_card(i)
        )
        button.grid(row=i // 4, column=i % 4, padx=5, pady=5)
        buttons.append(button)

    tk.Button(
        frame, text="🔄 NEW GAME", font=("Arial", 13, "bold"),
        bg="#E67E22", fg="white", width=15, command=reset
    ).pack(pady=10)


# =========================================================
# SUDOKU
# =========================================================

def sudoku():
    game = tk.Toplevel(root)
    game.title("🔢 Sudoku")
    game.geometry("650x750")
    game.resizable(False, False)

    set_background(game, "sudoku_bg.jpg", 650, 750)
    play_music("sudoku_music.mp3")
    create_music_button(game, "sudoku_music.mp3")

    game.protocol("WM_DELETE_WINDOW", lambda: close_game(game))

    frame = tk.Frame(game, bg="#17202A")
    frame.place(x=25, y=65, width=600, height=665)

    tk.Label(
        frame, text="🔢 SUDOKU", font=("Arial", 28, "bold"),
        fg="#F7DC6F", bg="#17202A"
    ).pack(pady=8)

    difficulty = tk.StringVar(value="Easy")

    difficulty_frame = tk.Frame(frame, bg="#17202A")
    difficulty_frame.pack(pady=5)

    tk.Label(
        difficulty_frame, text="Difficulty:", font=("Arial", 12, "bold"),
        fg="white", bg="#17202A"
    ).pack(side="left", padx=5)

    difficulty_menu = tk.OptionMenu(
        difficulty_frame, difficulty, "Easy", "Medium", "Hard"
    )
    difficulty_menu.config(font=("Arial", 11, "bold"), width=10)
    difficulty_menu.pack(side="left")

    board_frame = tk.Frame(frame, bg="#17202A")
    board_frame.pack(pady=5)

    entries = []
    puzzle = []
    solution = []

    def is_valid(board, row, col, num):
        for i in range(9):
            if board[row][i] == num or board[i][col] == num:
                return False

        start_row, start_col = (row // 3) * 3, (col // 3) * 3
        for i in range(start_row, start_row + 3):
            for j in range(start_col, start_col + 3):
                if board[i][j] == num:
                    return False
        return True

    def solve_board(board):
        for row in range(9):
            for col in range(9):
                if board[row][col] == 0:
                    numbers = list(range(1, 10))
                    random.shuffle(numbers)
                    for num in numbers:
                        if is_valid(board, row, col, num):
                            board[row][col] = num
                            if solve_board(board):
                                return True
                            board[row][col] = 0
                    return False
        return True

    def generate_sudoku():
        nonlocal puzzle, solution
        solution = [[0 for _ in range(9)] for _ in range(9)]
        solve_board(solution)

        puzzle = [row[:] for row in solution]
        remove_count = 35 if difficulty.get() == "Easy" else (45 if difficulty.get() == "Medium" else 55)

        positions = [(r, c) for r in range(9) for c in range(9)]
        random.shuffle(positions)

        for i in range(remove_count):
            r, c = positions[i]
            puzzle[r][c] = 0

    def display_puzzle():
        for row in range(9):
            for col in range(9):
                value = puzzle[row][col]
                entry = entries[row][col]
                entry.config(state="normal")
                entry.delete(0, tk.END)

                if value != 0:
                    entry.insert(0, str(value))
                    entry.config(state="disabled", disabledforeground="#D35400")
                else:
                    entry.config(bg="white", fg="black")

    def new_game():
        generate_sudoku()
        display_puzzle()

    def check_answer():
        correct = True
        for row in range(9):
            for col in range(9):
                value = entries[row][col].get()
                if value == "":
                    correct = False
                    entries[row][col].config(bg="#F5B7B1")
                else:
                    try:
                        if int(value) != solution[row][col]:
                            correct = False
                            entries[row][col].config(bg="#F5B7B1")
                        else:
                            entries[row][col].config(bg="#ABEBC6")
                    except ValueError:
                        correct = False
                        entries[row][col].config(bg="#F5B7B1")

        if correct:
            messagebox.showinfo("Sudoku", "🎉 Congratulations!\nYou solved the Sudoku!")

    def clear_answer():
        for row in range(9):
            for col in range(9):
                if puzzle[row][col] == 0:
                    entries[row][col].delete(0, tk.END)
                    entries[row][col].config(bg="white")

    def solve_current():
        for row in range(9):
            for col in range(9):
                entries[row][col].config(state="normal")
                entries[row][col].delete(0, tk.END)
                entries[row][col].insert(0, str(solution[row][col]))
                entries[row][col].config(bg="#ABEBC6")

    # Construct Grid
    for row in range(9):
        row_entries = []
        for col in range(9):
            pad_x = (2, 5) if (col + 1) % 3 == 0 and col < 8 else (2, 2)
            pad_y = (2, 5) if (row + 1) % 3 == 0 and row < 8 else (2, 2)

            entry = tk.Entry(
                board_frame, font=("Arial", 16, "bold"), width=2,
                justify="center", relief="solid", bd=1
            )
            entry.grid(row=row, column=col, padx=pad_x, pady=pad_y)
            row_entries.append(entry)
        entries.append(row_entries)

    btn_frame = tk.Frame(frame, bg="#17202A")
    btn_frame.pack(pady=10)

    tk.Button(
        btn_frame, text="🔄 New", font=("Arial", 11, "bold"),
        bg="#E67E22", fg="white", width=8, command=new_game
    ).grid(row=0, column=0, padx=5)

    tk.Button(
        btn_frame, text="✅ Check", font=("Arial", 11, "bold"),
        bg="#2ECC71", fg="white", width=8, command=check_answer
    ).grid(row=0, column=1, padx=5)

    tk.Button(
        btn_frame, text="🧹 Clear", font=("Arial", 11, "bold"),
        bg="#3498DB", fg="white", width=8, command=clear_answer
    ).grid(row=0, column=2, padx=5)

    tk.Button(
        btn_frame, text="💡 Solve", font=("Arial", 11, "bold"),
        bg="#9B59B6", fg="white", width=8, command=solve_current
    ).grid(row=0, column=3, padx=5)

    new_game()


# =========================================================
# MAIN MENU UI
# =========================================================

# Set Main Window Background
set_background(root, "main_bg.jpg", 700, 700)

# Title Banner
title_label = tk.Label(
    root,
    text="🎮  GAME COLLECTION  🎮",
    font=("Arial", 22, "bold"),
    fg=TITLE_COLOR,
    bg="#0B131E",
    padx=20,
    pady=10,
    relief="solid",
    bd=2
)
title_label.pack(pady=(25, 10))

# Subtitle Banner
subtitle_label = tk.Label(
    root,
    text="Choose a game and have fun!",
    font=("Arial", 13, "bold"),
    fg="white",
    bg="#0B131E",
    padx=15,
    pady=5
)
subtitle_label.pack(pady=(0, 20))

# Menu Buttons Setup
button_style = {
    "font": ("Arial", 13, "bold"),
    "fg": "white",
    "width": 26,
    "height": 1,
    "bd": 2,
    "relief": "raised",
    "cursor": "hand2"
}

tk.Button(
    root, text="❌ ⭕    TIC-TAC-TOE", bg="#E74C3C",
    command=tic_tac_toe, **button_style
).pack(pady=6)

tk.Button(
    root, text="🎲    SNAKE & LADDER", bg="#2980B9",
    command=snake_and_ladder, **button_style
).pack(pady=6)

tk.Button(
    root, text="✊    ROCK PAPER SCISSORS", bg="#8E44AD",
    command=rock_paper_scissors, **button_style
).pack(pady=6)

tk.Button(
    root, text="🔢    NUMBER GUESSING", bg="#27AE60",
    command=number_guessing, **button_style
).pack(pady=6)

tk.Button(
    root, text="🧠    MEMORY GAME", bg="#9B59B6",
    command=memory_game, **button_style
).pack(pady=6)

tk.Button(
    root, text="🔢    SUDOKU", bg="#16A085",
    command=sudoku, **button_style
).pack(pady=6)

tk.Button(
    root, text="🚪    EXIT", bg="#5D6D7E",
    command=root.quit, **button_style
).pack(pady=(6, 25))

# Footer Tag
footer_label = tk.Label(
    root,
    text="Made with Python 🐍",
    font=("Arial", 10, "bold"),
    fg="white",
    bg="#0B131E",
    padx=10,
    pady=3
)
footer_label.pack(side="bottom", pady=15)

# Start background music for main window
play_music("main_music.mp3")

root.mainloop()
```
